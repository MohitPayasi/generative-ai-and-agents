```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix,classification_report
from sklearn.model_selection import train_test_split
```

```python
df=sns.load_dataset('tips')
```

```python
df.isnull().sum()
```

```python
df['sex']= df['sex'].map({'Male':1,'Female':0})
```

```python
df['smoker']= df['smoker'].map({'Yes':1,'No':0})
```

```python
df.columns
```

```python
x=df[[ 'tip','size']]
y=df['sex']
```

```python
x_train, x_test, y_train, y_test= train_test_split(
    x,y,test_size=0.2,random_state=42
)
```

```python
log_res= LogisticRegression()
```

```python
log_res.fit(x_train,y_train)
```

```python
y_pred=log_res.predict(x_test)
y_pred
```

```python
confusion_matrix(y_test,y_pred)
```

```python
print(classification_report(y_pred,y_test))
```

when precision and recall both are important

```python
import numpy as np
y_prob=log_res.predict_proba(x_test)[:,-1]
```

```python
thresold=np.arange(0,1,0.1)
```

```python
from sklearn.metrics import recall_score, precision_score,f1_score
```

```python
result=[]
for  vth in thresold:
    y_pred= (y_prob>vth).astype(int)
    precision_score_1= precision_score(y_test,y_pred)
    recall_score_1= recall_score(y_test,y_pred)
    f1_score_1= f1_score(y_test,y_pred,zero_division=0)
    result.append([vth,precision_score_1,recall_score_1,f1_score_1])
```

```python
df=pd.DataFrame(result,columns=['vth','precision', 'recall', 'f1_score'])
```

```python
row=df.loc[df['f1_score'].idxmax()]
row['vth']
```

```python
thresold
```

```python
specificity= 1-fpr

best_value= tpr+specificity-1
best_index= np.argmax(best_value)
thresold[best_index]
```
