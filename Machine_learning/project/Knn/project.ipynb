## Random UnderSampling

```python
x=df.drop('Loan_Status', axis=1)
y= df['Loan_Status']
```

```python
x
```

```python
from imblearn.under_sampling import RandomUnderSampler
from collections import Counter
```

```python
Counter(y)
```

```python
rus=RandomUnderSampler(sampling_strategy='auto',
random_state=42)
```

```python
x_under,y_under=rus.fit_resample(x,y)
Counter(y_under)
```

```python
from imblearn.over_sampling import RandomOverSampler
ros= RandomOverSampler(sampling_strategy='auto',
random_state=42)
x_over, y_over=ros.fit_resample(x,y)
Counter(y_over)
```

```python
from imblearn.over_sampling import SMOTE
smt= SMOTE(sampling_strategy='auto', random_state=42, k_neighbors=5)
x_smote,y_smote=smt.fit_resample(x,y)
Counter(y_smote)
```

```python
from imblearn.over_sampling import ADASYN
ada= ADASYN(sampling_strategy='auto', random_state=42)
x_ada,y_ada=ada.fit_resample(x,y)
Counter(y_ada)
```

```python
from imblearn.under_sampling import TomekLinks
tmk= TomekLinks(sampling_strategy='auto')
x_tmk, y_tmk=tmk.fit_resample(x,y)
Counter(y_tmk)
```

```python
from imblearn.under_sampling import NearMiss
nrmss= NearMiss(sampling_strategy='auto')
x_nrms, y_nrms=nrmss.fit_resample(x,y)
Counter(y_nrms)
```
