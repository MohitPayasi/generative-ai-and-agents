```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
os.environ['GROQ_API_KEY']= os.getenv('GROQ_API_KEY')
```

```python
from langchain_community.document_loaders import WebBaseLoader
from langchain_community.vectorstores import FAISS
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
urls=['https://plato.stanford.edu/entries/ethics-virtue/',
        'https://charactercounts.org/','https://www.psychologytoday.com/us/basics/empathy']
```

```python
docs= [WebBaseLoader(url).load() for url in urls]
docs
```

```python
docs_list=[doc for sublist in docs for doc in sublist]
```

```python
docs_list
```

```python
text_splitter= RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=200)
doc_splits=text_splitter.split_documents(docs_list)
print(doc_splits)
```

```python
os.environ['HF_TOKEN'] = os.getenv('HF_TOKEN')
from langchain_huggingface import HuggingFaceEmbeddings
embedding=HuggingFaceEmbeddings(model_name='all-MiniLM-L6-v2')
```

Ollama v0.40.18! 

```python
vector_store=FAISS.from_documents(documents=doc_splits, embedding=embedding)
```

```python
retriver= vector_store.as_retriever()
```
