```python
from pydantic import BaseModel,Field
```

```python
from langgraph.graph import StateGraph,END,START
```

```python
class State(BaseModel):
    name:str

```

```python
def example_node(state:State):
    return {'name':'Hello'}
```

```python
builder= StateGraph(State)
builder.add_node('node_1',example_node)
builder.add_edge(START,'node_1')

builder.add_edge('node_1',END)
```

```python
builder.compile()
```

```python
builder.compile().invoke({'name':'mohit'})
```

```python
builder.compile().invoke({'name':123}) # we can not enter other data type
```
