```python
import os
from typing_extensions import TypedDict
from langgraph.graph import StateGraph,START,END
from typing import Literal
import random
# reducer
from langgraph.graph.message import add_messages
from typing import Annotated
os.environ['GROQ_API_KEY']= os.getenv('GROQ_API_KEY')
from langchain.chat_models import init_chat_model
llm= init_chat_model(model='openai/gpt-4.1',model_provider='groq')
```

```python
class TypedDictstate(TypedDict):
    name:str
    game:Literal['cricket','badminton']
    
```

```python
def play_game(state:TypedDictstate):
    print('---Play Game Node has been called--')
    return {'name':state['name']+ f' want to play {state['game']}'}

def cricket(state:TypedDictstate):
    print('cricket node has been called')

    return {'game':'cricket'}
def badminton(state:TypedDictstate):
    print('badminton node has been called')

    return {'game':'badminton'}

def decide_play(state:TypedDictstate)->Literal['cricket','badminton']:
    if random.random()<0.5:
        return 'cricket'
    else:
        return 'badminton'



graph= StateGraph(TypedDictstate)

graph.add_node('node_1',play_game)
graph.add_node('node_2',cricket)
graph.add_node('node_3',badminton)

graph.add_edge(START,'node_1')

graph.add_conditional_edges("node_1", decide_play, {"cricket": "node_2", "badminton": "node_3"})

graph.add_edge('node_2',END)
graph.add_edge('node_3',END)
```

```python
graph=graph.compile()
```
