In LangChain, a Message represents a unit of communication between a user, an AI model, or a tool.

Every message generally contains two important things:
1. Role: Who is sending the message?
2. Content: What is the message saying?
1. What is a Role in LangChain?
A role defines the identity or responsibility of the message sender.


LangChain commonly uses these message roles:
Role	Meaning	Purpose
system	SystemMessage	Defines AI behaviour and instructions

human	HumanMessage	Represents input from the user

ai	AIMessage	Represents the AI model's response

tool	ToolMessage	Represents the output returned by a tool



2. What is User in LangChain?
The user is the person interacting with the AI application.


In LangChain, user messages are generally represented by the HumanMessage class, which uses the human role.

Example:
from langchain_core.messages import HumanMessagemessage = HumanMessage(    content="What is Machine Learning?")print(message)


Output:
HumanMessage(
    content='What is Machine Learning?'
)

What is a Text Prompt in LangChain?
A Text Prompt is an instruction or input given to an AI model to generate a response.

In simple words:

Prompt = Instructions + User Input + Context

For example:
Explain Machine Learning in simple language with an example.


This complete sentence is a text prompt.
1. Text Prompt vs Message
Text Prompt	Message

Instruction or input given to an AI	Structured communication
Usually plain text or a template	Contains role and content
Can contain variables	Can represent system, human, AI or tool
Example: Explain Python	Example: HumanMessage("Explain Python")
