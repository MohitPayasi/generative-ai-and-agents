Structured Output in LangChain allows an LLM to generate responses according to a predefined schema. It makes AI responses predictable, machine-readable, and easier to integrate with Python applications, APIs, databases, and AI agents.


Structured Output means instructing an LLM to return its response in a predefined format or structure instead of generating unrestricted text.


In simple words:


Normal output gives you a paragraph. Structured output gives you data in a predictable format, such as JSON, a Python dictionary, or a Pydantic object.

```python
from pydantic import BaseModel, Field

import os
from dotenv import load_dotenv
load_dotenv()
from langchain.chat_models import init_chat_model
os.environ['GROQ_API_KEY']= os.getenv('GROQ_API_KEY')
```

```python
from langchain.chat_models import init_chat_model

llm = init_chat_model(
    model="openai/gpt-oss-120b",
    model_provider="groq",
    temperature=0
)
```

```python
from pydantic import BaseModel, Field

class Movie(BaseModel):
    title: str = Field(description="Movie title")
    release_year: int = Field(description="Release year")
    director: str = Field(description="Director name")
    genre: str = Field(description="Movie genre")
    language: str = Field(description="Movie language")
```

```python
structured_llm = llm.with_structured_output(
    Movie,
    method="json_schema")

response = structured_llm.invoke(
    "Tell me about Bajrangi Bhaijaan")

print(response)
```

```python
from pydantic import BaseModel, Field

class Movie(BaseModel):
    title: str = Field(...,description="Movie title")
    release_year: int = Field(...,description="Release year")
    director: str = Field(...,description="Director name")
    genre: str = Field(...,description="Movie genre")
    language: str = Field(...,description="Movie language")
```

```python
structured_llm = llm.with_structured_output(
    Movie,
    method="json_schema")

response = structured_llm.invoke(
    "Tell me about Bajrangi Bhaijaan")

print(response)
```

```python
class Actor(BaseModel):
    name:str
    age:int
    role:str

class Movie(BaseModel):
    title: str = Field(...,description="Movie title")
    release_year: int = Field(...,description="Release year")
    director: str = Field(...,description="Director name")
    cast:list[Actor]
    genre: str = Field(...,description="Movie genre")
    language: str = Field(...,description="Movie language")
    budget : float |None= Field(None,discription='Budget in millions USD')
```

```python
structured_llm = llm.with_structured_output(
    Movie,
    method="json_schema")

response = structured_llm.invoke(
    "Tell me about Avatar")

print(response)
```

"TypedDict is used to define the expected structure and types of dictionary keys, mainly for static type checking. It does not perform runtime validation.


Pydantic is a data validation library that validates data at runtime, supports type conversion, and raises validation errors when the input data is invalid.


In short, TypedDict defines the expected structure, whereas Pydantic validates the actual data."

```python
from typing_extensions import TypedDict,Annotated
```

```python
from typing_extensions import TypedDict

from typing_extensions import TypedDict, Annotated

class Movie(TypedDict):
    title: Annotated[str, "The name of the movie"]
    year: Annotated[int, "The release year of the movie"]
    director: Annotated[str, "The name of the movie director"]
```

```python
structured_llm = llm.with_structured_output(
    Movie,
    method="json_schema")

structured_llm.invoke('please provide details of  avatar')
```

What is a Data Class?
A Data Class is used to create classes with less boilerplate code. It automatically generates useful methods such as:
- __init__() – Initializes object attributes.
- __repr__() – Provides a readable object representation.
- __eq__() – Compares objects.

Concept	Best Use Case
TypedDict	Dictionary structures and LangGraph State

Dataclass	Creating lightweight Python objects

Pydantic	Data validation, API schemas and LangChain structured output

Annotated	Adding metadata and descriptions to types

Remember these four definitions for AI Engineering interviews:
1. TypedDict: Defines the expected structure of a dictionary.
2. Dataclass: Simplifies the creation of Python classes and objects.
3. Pydantic: Validates and parses data at runtime.
4. Annotated: Adds metadata to existing type annotations.
