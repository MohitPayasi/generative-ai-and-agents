```python
import os
from dotenv import load_dotenv
load_dotenv()
os.environ['GROQ_API_KEY']= os.getenv('GROQ_API_KEY')
from langchain.chat_models import init_chat_model
llm= init_chat_model(model='openai/gpt-oss-20b', model_provider='groq')
```

"Invoke, Stream, and Batch are three execution methods in LangChain. 

Invoke processes a single input and returns the complete output. 

Stream returns the output incrementally as it is generated, improving the user experience in real-time applications. 


Batch processes multiple independent inputs and returns their corresponding outputs, making it useful for bulk processing and large-scale LLM applications."

```python
for chunk in llm.stream('tell me about rocket science'):
    print(chunk.text,end=' ')
```

```python
responce=llm.batch(['what is streaming',
'where we can use streaming',
'what is batch in lanchain'], config={'max_concurrency':10}
)
for chunk in responce:
    print(chunk.text, end=' ', )
```
