"Invoke, Stream, and Batch are three execution methods in LangChain. 

Invoke processes a single input and returns the complete output. 

Stream returns the output incrementally as it is generated, improving the user experience in real-time applications. 


Batch processes multiple independent inputs and returns their corresponding outputs, making it useful for bulk processing and large-scale LLM applications."
