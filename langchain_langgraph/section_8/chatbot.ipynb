```python
store = {}
def get_session_history(session_id: str) -> BaseChatMessageHistory:
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]
```

### ChatPromptTemplate with MessagesPlaceholder
**Purpose:** Ye prompt template system message define karta hai aur `MessagesPlaceholder` ke zariye dynamic chat history/messages accept karta hai.

- **Inputs:**
  - `system`: Instructions for assistant behavior
  - `messages`: List of `HumanMessage` / `AIMessage` history passed dynamically
- **Output:** Formatted chat prompt jo `prompt | llm` chain me feed hota hai.


```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import HumanMessage

# ChatPromptTemplate.from_messages expects a LIST of messages: [ ... ]
prompt = ChatPromptTemplate.from_messages([
    ("system", "you are helpfull assistant answer all question your best knowledge"),
    MessagesPlaceholder(variable_name="messages")
])

chain = prompt | llm
response = chain.invoke({"messages": [HumanMessage(content="my name is mohit payasi")]})
response
```

### Clear Chat Session History (`store.clear()`)
**Purpose:** In-memory dictionary `store` se saari purani conversations aur chat history delete karne ke liye taaki chatbot fresh memory ke saath start kare.

- **Logic:**
  - `store.clear()`: Saare sessions (`session_id`) ki saved message history ko memory se wipe out karta hai.
  - `store[session_id].clear()`: Kisi ek specific session ki message history ko clear karta hai.
- **Return:** `None` (Memory dictionary empty ho jaati hai).


```python
# Saari sessions ki chat history delete karne ke liye:
store.clear()

# Ya kisi specific session ki history delete karne ke liye:
# if 'session_1' in store:
#     store['session_1'].clear()

print("Chat history cleared successfully! Active sessions:", len(store))
```
